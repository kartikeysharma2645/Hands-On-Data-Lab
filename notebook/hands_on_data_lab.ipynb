### Practical Numerical Data Operation

NumPy can be used to perform condition-based operations on numerical data. In this exercise, passenger ages are analyzed using Boolean indexing to identify values above the average age.

```python
ages = np.array([12, 18, 25, 31, 42, 55, 67, 8])
avg_age = np.mean(ages)
print("Average age:", avg_age)

max_age = np.max(ages)
print("Maximum age:", max_age)

min_age = np.min(ages)
print("Minimum age:", min_age)

print("\nAges greater than average age:", ages[ages > avg_age])  # Filter ages greater than average age
```

### Missing Age Analysis

The distribution of age is examined across passenger class and sex to determine an appropriate strategy for handling missing age values.

```python
# Fill missing age values with the median age for each passenger class and sex
df.groupby(["Pclass", "Sex"])["Age"].median()
```

```python
# Fill missing age values with the median age for each passenger class and sex
df["Age"] = df.groupby(["Pclass", "Sex"])["Age"].transform(
    lambda x: x.fillna(x.median())
    )

df["Age"].isnull().sum()
```

### Handling Missing Embarkation Values

Missing values in the categorical `Embarked` column are handled using the most frequent category after examining its distribution.

```python
# Fill missing embarked values with the most common value
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df["Embarked"].isnull().sum()
```

### Cabin Missingness Analysis

The extent of missing values in the `Cabin` column is examined before deciding whether the column should be retained, transformed, or excluded from the analysis.

```python
# check for missing values in the "Cabin" column
df["Cabin"].isnull().sum()
```

```python
# Calculate the percentage of missing values in the "Cabin" column
missing_cabin_pct = df["Cabin"].isnull().sum() / len(df) * 100
print("Missing Cabin percentage:", missing_cabin_pct)
```

```python
# Remove the "Cabin" column from the dataset
cleaned_df = df.copy()
cleaned_df = cleaned_df.drop(columns=["Cabin"])

print("Original shape:", df.shape)
print("Cleaned shape:", cleaned_df.shape)
```

```python
# Check for missing values in the cleaned DataFrame
cleaned_df.isnull().sum()
```

```python
# Compare the shapes of the original and cleaned datasets
print("Original dataset shape:", df.shape)
print("Cleaned dataset shape:", cleaned_df.shape)

print("\nRemoved columns:")
print(set(df.columns) - set(cleaned_df.columns))
```

```python
# Create a new column "FamilySize" by adding "SibSp" and "Parch" columns
cleaned_df["FamilySize"] = cleaned_df["SibSp"] + cleaned_df["Parch"] + 1

cleaned_df[["SibSp", "Parch", "FamilySize"]].head()
```

```python
# Display the value counts of the "FamilySize" column in the cleaned DataFrame
cleaned_df["FamilySize"].value_counts().sort_index()
```

```python
# Create a new column "FamilyType" based on the "FamilySize" column
conditions = [
    cleaned_df["FamilySize"] == 1,
    cleaned_df["FamilySize"].between(2, 4),
    cleaned_df["FamilySize"] >= 5
]

choices = ["Alone", "Small Family", "Large Family"]

cleaned_df["FamilyType"] = np.select(
    conditions,
    choices,
    default="Unknown"
)
```

```python
# Display the first 10 rows of the "FamilySize" and "FamilyType" columns
cleaned_df[["FamilySize", "FamilyType"]].head(10)
```

```python
# Display the value counts of the "FamilyType" column in the cleaned DataFrame
cleaned_df["FamilyType"].value_counts()
```
