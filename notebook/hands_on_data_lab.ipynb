## NumPy

NumPy provides efficient array-based operations for numerical computing in Python.

```python
# Create a 1D NumPy array
array = np.array([1, 2, 3, 4, 5, 6])

array.ndim
```

```python
# Create a 1D NumPy array
array.shape
```

```python
# Get the data type of the array
array.dtype
```

```python
array[0]  # Access the first element of the array
```

```python
# Access the last element of the array
array[-1]
```

```python
# Access elements from index 1 to 3
array[1:4]  
```

```python
# Reverse the array
array[::-1]  
```

```python
# Create a 2D NumPy array
array2 = np.array([[1, 2, 3], [4, 5, 6]])

array2.shape
```

```python
# Reshape the array to the specified number of rows and columns
reshaped_array = array.reshape(2, 3)

reshaped_array.shape
```

```python
print("Original array:",array)
print("\nReshaped array:\n",reshaped_array)
print("\nShape of the reshaped array:",reshaped_array.shape)
```

```python
np.sum(array)  # Calculate the sum of all elements in the array
```

```python
# Calculate the mean of all elements in the array
np.mean(array)
```

```python
# Calculate the standard deviation of all elements in the array
np.std(array)
```
